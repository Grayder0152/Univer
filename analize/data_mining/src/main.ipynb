```python
from clustering.utils.visualizer import Visualizer2DNoClustered, Visualizer2D
from clustering import DistanceMethodName, KMeans, KMedoids
from extranction import extract
```

## Data extraction

```python
df = extract('data_2.csv')
Visualizer2DNoClustered().plot(df)
```

## Clustering

### K-Mean

```python
clustered_df, centroids = KMeans(3, distance_method_name=DistanceMethodName.EUCLIDEAN).clustering(df)
Visualizer2D().plot(clustered_df, feature_names = ['col1', 'col2'], centroids = centroids)
```

### K-medoid

```python
clustered_df, medoids = KMedoids(3, distance_method_name=DistanceMethodName.MANHATTAN).clustering(df)
Visualizer2D().plot(clustered_df, feature_names = ['col1', 'col2'], centroids = medoids)
```
